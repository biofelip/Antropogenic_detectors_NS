```python
from opensoundscape.visualization import annotate
import numpy as np
import pandas as pd
from glob import glob
from pathlib import Path
import matplotlib.pyplot as plt
import opensoundscape as opso
import bioacoustics_model_zoo as bmz
import torch, torchaudio
from opensoundscape import Audio


# load the catalog for the test set
test_set_df = pd.read_csv("test_set/catalog_sample.csv")
airgun_files = test_set_df[test_set_df["group"] == "airgun"]["path"].values
other_files = test_set_df[test_set_df["group"] == "other"]["path"].values

```

```python

clips_ag = opso.utils.make_clip_df(airgun_files, clip_duration=5)
clips_ot = opso.utils.make_clip_df(other_files, clip_duration=5)

```

```python
clips_ag_df = clips_ag.reset_index()
clips_ag_df["airgun"] = None
```

```python

start_idx = 300
print(f"Annotating clips {start_idx} to {start_idx+50} of {len(clips_ag_df)}")
idx = clips_ag_df.index[start_idx:start_idx+2]
widget= annotate(clips_ag_df,
                 indices=idx,
                 N=50,
                 annotation_buttons=['airgun'],bandpass_range=(0,2000),
                 spec_kwargs={"window_samples": 8200, "overlap_fraction": 0.75})
```

```python
# save the airgun labels to a csv file
test_df_ag = clips_ag_df.set_index(["file", "start_time", "end_time"])[["airgun"]].copy()
test_df_ag["airgun"] = test_df_ag["airgun"].eq(True).astype(int)     # clicked → 1, not clicked → 0
print(test_df_ag["airgun"].value_counts())
test_df_ag.to_csv("test_set/test_set_labels_ag.csv")
```

```python
# if stuff has been annotated already load it from memory dont create the clips dataframe again
clips_ot_df = pd.read_csv("test_set/test_set_labels_ot.csv")
```

```python
clips_ot_df = clips_ot.reset_index()
clips_ot_df["airgun"] = None
```

```python
#batches with animals 2750 3200
start_idx = 3950
print(f"Annotating clips {start_idx} to {start_idx+50} of {len(clips_ot_df)}")
idx = clips_ot_df.index[start_idx:start_idx+50]
widget= annotate(clips_ot_df,
                 indices=idx,
                 N=50,
                 annotation_buttons=['airgun', "minke?"],bandpass_range=(0,2000),
                 spec_kwargs={"window_samples": 8200, "overlap_fraction": 0.75})
```
